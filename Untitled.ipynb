```python
threshold = len(df) * 0.60
df2 = df.dropna(thresh=threshold, axis=1) 
```

```python
df6 = pd.get_dummies(df2,columns=['WindGustDir','WindDir9am','WindDir3pm'],dtype=float)
df6
```

```python
from sklearn.preprocessing import LabelEncoder

df4 = df6.copy()
encoders = {}
for col in ['RainToday', 'RainTomorrow']:      
    encoders[col] = LabelEncoder()
    df4[col] = encoders[col].fit_transform(df4[col])

df4
```

```python
df4.loc[0:20,'RainTomorrow']
```

```python
df4.info()
```

```python

cols = df4.iloc[:, 2:17].select_dtypes(include='number').columns
print(cols.tolist())
```

```python
Q1 = df4.groupby('Location')[cols].transform(lambda x: x.quantile(0.25))
Q3 = df4.groupby('Location')[cols].transform(lambda x: x.quantile(0.75))
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
df4
```

```python

mask = (df4[cols] < lower) | (df4[cols] > upper)
df4_clean = df4[~mask.any(axis=1)]
```

```python
df4
```

```python
df4_clean.info()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler


cols = ['MinTemp', 'MaxTemp', 'Rainfall', 'WindGustSpeed', 'WindSpeed9am',
        'WindSpeed3pm', 'Pressure9am', 'Pressure3pm', 'Cloud9am', 'Temp9am', 'Temp3pm']

ct = ColumnTransformer(
    transformers=[('scaler', MinMaxScaler(), cols)],
    remainder='passthrough'
)

df4_clean = pd.DataFrame(ct.fit_transform(df4_clean))
```

```python
df4_clean
```

```python
from sklearn.model_selection import train_test_split
df4_clean_train, df4_clean_test = train_test_split( df4_clean, test_size = 0.2, random_state = 1)
```
